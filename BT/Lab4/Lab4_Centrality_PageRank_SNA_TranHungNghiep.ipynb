```python
import networkx as nx
import matplotlib.pyplot as plt

def topk(d, k=5):
    return sorted(d.items(), key=lambda x: x[1], reverse=True)[:k]
```

## 1) Toy Graph
Create a small undirected graph and visualize it.

```python
edges = [
    ("Alice","Bob"),("Alice","Carol"),("Bob","Carol"),
    ("Carol","Dave"),("Dave","Eve"),("Eve","Frank"),
    ("Eve","Grace"),("Grace","Heidi"),("Frank","Heidi"),
    ("Ivan","Judy"),("Ivan","Alice"),("Judy","Carol")
]
G = nx.Graph(edges)

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())

pos = nx.spring_layout(G, seed=42)
plt.figure(figsize=(8,6))
nx.draw(G, pos, with_labels=True, node_color="lightblue", node_size=800, font_size=10)
plt.title("Toy Graph")
plt.axis("off")
plt.show()
```

## 2) Centrality Computations
Compute multiple centrality measures and list top-5 nodes for each.

```python
# Degree
deg = nx.degree_centrality(G)
# Closeness
clo = nx.closeness_centrality(G)
# Betweenness
bet = nx.betweenness_centrality(G)
# Eigenvector (may need more iterations on some graphs)
eig = nx.eigenvector_centrality(G, max_iter=1000)
# PageRank (alpha = 0.85)
pr = nx.pagerank(G, alpha=0.85)

print("Top Degree:", topk(deg))
print("Top Closeness:", topk(clo))
print("Top Betweenness:", topk(bet))
print("Top Eigenvector:", topk(eig))
print("Top PageRank:", topk(pr))
```

## 3) Visualize by Centrality
Color and size nodes by PageRank (you can switch to any measure by changing the `values` list).

```python
# Question: also try visualizing other centrality measures, compare between them visually.

plt.figure(figsize=(8,6))
values = [pr[n] for n in G.nodes()]
sizes = [4000*v for v in values]
nx.draw(G, pos, with_labels=True, node_color=values, cmap=plt.cm.viridis, node_size=sizes, font_size=10)
plt.title("Toy Graph: Node size/color = PageRank")
plt.axis("off")
plt.show()
```

## 4) Real Dataset: Karate Club
Run the same analyses on the classic Karate Club social network.

```python
G2 = nx.karate_club_graph()
pos2 = nx.spring_layout(G2, seed=123)

pr2 = nx.pagerank(G2, alpha=0.85)
bet2 = nx.betweenness_centrality(G2)
eig2 = nx.eigenvector_centrality(G2, max_iter=1000)
deg2 = nx.degree_centrality(G2)

print("Top 5 by PageRank:", topk(pr2))
print("Top 5 by Betweenness:", topk(bet2))
print("Top 5 by Eigenvector:", topk(eig2))
print("Top 5 by Degree:", topk(deg2))

plt.figure(figsize=(8,6))
values2 = [pr2[n] for n in G2.nodes()]
sizes2 = [4000*v for v in values2]
nx.draw(G2, pos2, with_labels=True, node_color=values2, cmap=plt.cm.plasma, node_size=sizes2, font_size=8)
plt.title("Karate Club: Node size/color = PageRank")
plt.axis("off")
plt.show()
```

## 5) Teleportation Sensitivity in PageRank
Compare rankings under different `alpha` (teleport probability = `1 - alpha`).

```python
for a in [0.50, 0.85, 0.95]:
    pr_a = nx.pagerank(G2, alpha=a)
    print(f"Alpha={a}: Top 5 by PageRank:", topk(pr_a))
```
